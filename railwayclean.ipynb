```python
import random

class RailwayStation:
    def __init__(self):
        self.locationCondition = {}
        self.crowdCondition = {}

        n = int(input("Enter number of station locations: "))

        for i in range(n):
            location = input("Enter location name: ")

            condition = int(
                input("Enter condition (0 = Clean, 1 = Dirty): ")
            )

            crowd = int(
                input("Is location crowded? (0 = No, 1 = Yes): ")
            )

            self.locationCondition[location] = condition
            self.crowdCondition[location] = crowd


class CleaningRobot:
    def __init__(self, Environment):

        print("\nInitial Station Environment:")
        print(Environment.locationCondition)

        Score = 0
        Battery = 100
        Dustbin = 0

        # Random starting location
        locations = list(Environment.locationCondition.keys())
        robotLocation = random.choice(locations)

        print("\nRobot is placed at:", robotLocation)

        # Visit all locations
        for location in locations:

            # Move to location
            if location != robotLocation:

                print("\nMoving to", location)

                Score -= 1
                Battery -= 5

                robotLocation = location

            # Check crowd
            if Environment.crowdCondition[location] == 1:

                print("Location", location, "is Crowded.")
                print("Robot is waiting...")

                Battery -= 2
                continue

            # Check dirty condition
            if Environment.locationCondition[location] == 1:

                print("Location", location, "is Dirty.")

                # Clean location
                Environment.locationCondition[location] = 0

                Score += 1
                Battery -= 10
                Dustbin += 20

                print("Location", location, "has been Cleaned.")

            else:

                print("Location", location, "is Clean.")

        # Final result
        print("\nFinal Station Environment:")
        print(Environment.locationCondition)

        print("\nPerformance Measurement:", Score)
        print("Battery Remaining:", Battery, "%")
        print("Dustbin Filled:", Dustbin, "%")


# Create railway station
theEnvironment = RailwayStation()

# Create cleaning robot
theRobot = CleaningRobot(theEnvironment)
```
